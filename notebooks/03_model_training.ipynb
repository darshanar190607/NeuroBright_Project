# NeuroBright: Model Training

Train EEGNet model on preprocessed EEG data.

```python
import sys
sys.path.insert(0, '..')

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.utils.logger import get_logger, log_section_header
from src.utils.signal_utils import load_config
from src.training.trainer import ModelTrainer

sns.set_style('whitegrid')
print("✓ Libraries imported")
```

```python
# Load configuration
config = load_config()
logger = get_logger(__name__)

print("Training Configuration:")
print(f"  Epochs: {config['training']['epochs']}")
print(f"  Batch Size: {config['training']['batch_size']}")
print(f"  Learning Rate: {config['training']['learning_rate']}")
print(f"  Validation Split: {config['training']['val_split']}")
```

```python
# Initialize trainer
log_section_header("Model Training")
trainer = ModelTrainer(config)

print("✓ Trainer initialized")
```

```python
# Train model
history = trainer.train()

print(f"\n✓ Training complete!")
print(f"Best validation accuracy: {trainer.best_val_acc:.2f}%")
```

```python
# Plot training history
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Loss
ax1.plot(history['train_loss'], label='Train Loss', linewidth=2)
ax1.plot(history['val_loss'], label='Val Loss', linewidth=2)
ax1.set_title('Training and Validation Loss', fontsize=14, fontweight='bold')
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Loss')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Accuracy
ax2.plot(history['train_acc'], label='Train Accuracy', linewidth=2)
ax2.plot(history['val_acc'], label='Val Accuracy', linewidth=2)
ax2.set_title('Training and Validation Accuracy', fontsize=14, fontweight='bold')
ax2.set_xlabel('Epoch')
ax2.set_ylabel('Accuracy (%)')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Display final metrics
print("\nFinal Training Metrics:")
print(f"  Train Loss: {history['train_loss'][-1]:.4f}")
print(f"  Train Accuracy: {history['train_acc'][-1]:.2f}%")
print(f"  Val Loss: {history['val_loss'][-1]:.4f}")
print(f"  Val Accuracy: {history['val_acc'][-1]:.2f}%")
print(f"  Best Val Accuracy: {trainer.best_val_acc:.2f}%")
```
