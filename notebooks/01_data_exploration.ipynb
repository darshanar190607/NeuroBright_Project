Explore the EEG dataset structure, configuration, and project setup.

```python
import sys
sys.path.insert(0, '..')

from src.utils.signal_utils import load_config
import yaml
from pathlib import Path

print("✓ Libraries imported")
```

```python
# Load and display configuration
config = load_config()

print("NeuroBright Configuration:\n")
print(yaml.dump(config, default_flow_style=False))
```

```python
# Check directory structure
print("Project Structure:\n")

dirs = [
    config['paths']['raw_data'],
    config['paths']['processed_data'],
    config['paths']['saved_models'],
    config['paths']['logs']
]

for dir_path in dirs:
    p = Path(dir_path)
    exists = "✓" if p.exists() else "✗"
    print(f"{exists} {dir_path}")
```

```python
# Display brain states and recording protocol
print("Brain States to Record:\n")

for state_name, state_config in config['recording']['states'].items():
    print(f"{state_name.upper()} (Label {state_config['label']})")
    print(f"  Duration: {state_config['duration_seconds']}s")
    print(f"  Instruction: {state_config['instruction']}")
    print()
```

```python
# Display model architecture
print("EEGNet Model Configuration:\n")

model_config = config['model']
for key, value in model_config.items():
    print(f"  {key}: {value}")
```
